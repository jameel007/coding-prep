```python
import array
arr = array.array('i',[0,1,2,3,4])
print(arr)

##manipulating arrays
arr[0] = 10
print(arr)
```

```python
## Linear Search

def linear_search(arr,target):

    
    size = len(arr)
    for index in range(0,size):
        if arr[index] == target:
            return index
    return -1

lst = [0,10,20,25,60,320]

print(linear_search(lst,10))
```

```python
## Binary Search
def binary_search(arr,target):
   
    size = len(lst)
    start = 0
    end = size - 1
   ## print(len(arr))
    while (start<=end):
        mid = (end + start)//2
       
        if arr[mid] == target:
            return mid       
        
        elif arr[mid]>target:
            end = mid -1     
        
        elif arr[mid]<target:
            start= mid + 1 
    return -1

lst = [10,20,30,40,50,60,70,100]
target = 0
result = binary_search(lst,target)
print(result)
```
