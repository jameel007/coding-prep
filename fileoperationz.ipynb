## File Op R and W

```python
## Read a whole File

with open('sample2.txt', 'r') as file:
    content = file.read()
    print(content)
```

```python
## Read a file line by line

with open('sample2.txt','r') as file:
    for line in file:
        print(line)
        ## print(line.strip())  ##strip() removes the newline character
        
        
```

```python
## Writing a file and overwriting

with open('sample2.txt','w') as file:
    file.write('Hello World!\n')
    file.write('You will be releived.')  
## file will be overwritten, so solve use append 'a' instead of 'w'
```

```python
## Writing a file without overwriting
with open('sample2.txt','a') as file:
    file.write('you will be relieved!\n')
```

```python
## Writing a list of lines to a file

lines = ['First line\n','Second line\n','Third line\n']
with open('sample2.txt','a') as file:
    file.writelines(lines) 
```

```python
### Binary Files
## writing a binary file

data = b'\x00\x01\x03\x04'
with open('binary.bin','wb') as file:
    file.write(data)
```

```python
## Reading a binary file
with open('binary.bin','rb') as file:
    content = file.read()
    print(content)
```

```python
## Read the content from source and write to destination txt file
## copying a file

with open('sample2.txt', 'r') as source_file:
    content = source_file.read()

with open('destin_sample2.txt', 'w') as destination_file:
    destination_file.write(content)
```

```python
## Read a file and count the number of lines,word and characters.
def count_text_file(filepath):
    with open(filepath,'r') as file:
        lines = file.readlines()
        lines_count = len(lines)
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)
    return lines_count,word_count,char_count

filepath = "sample2.txt"
lines_count,word_count,char_count = count_text_file(filepath)    ## this is like c = add(3,4) but with multiple return values and since its in a tuple it is called unpacking a tuple.
print(f"Lines:{lines_count}, Words: {word_count},chars: {char_count}")
print(count_text_file(filepath))  ## tuple format
```
