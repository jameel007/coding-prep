```python
try:
    10/10
    a = b
except ZeroDivisionError as ex:
    print(ex)
    print("Cant bro")
except Exception as exx:  ## This is Base class exception 
    print(exx)            ## should be used at the end to catch everything below defined exceptions.
    
```

```python
try:
    num = int(input("Enter a number"))
    result = 10/num
except ValueError:
    print("Not a valid number")
except ZeroDivisionError:
    print("Enter > 0")
except Exception as ex:
    print(ex)
```

```python
# Try ,except,else block
try:
    num = int(input("Enter a number"))
    result = 10//num
except ValueError:
    print("Not a valid number")
except ZeroDivisionError:
    print("Enter > 0")
except Exception as ex:
    print(ex)
else:
    print("The result is:",result)  ## Executes after successfull execution of Try block.
finally:
    print("Execution Completed.") ## Executes regardless of try and catch.
```

## Try - tries the logic
## Except - exception if try fails
## else - Executes if try is successfull
## Finally - Executes regardless of try and except, else
