```python
def generate_hollow_square(n):
    """
    Function to return a hollow square pattern of '*' of side n as a list of strings.
    
    Parameters:
    n (int): The size of the square.
    
    Returns:
    list: A list of strings where each string represents a row of the hollow square.
    """
    
    square = []                                          
    if n == 1:                                            # Handle the case when n is 1 separately
        return ['*']                                      # The first and last rows are full of '*'
    square.append('*' * n)                                # First row
    for i in range(n - 2):                                # For the middle rows, the first and last characters are '*', rest are spaces
        square.append('*' + ' ' * (n - 2) + '*')
    square.append('*' * n)                                # Last row
    return square                                         # Return the list of rows
generate_hollow_square(5)  
```

```python
def reverse_pyramid(n):
##n = int(input("Enter the number,bitch:"))
    for i in range(n):
        print("*"*(n-i))
    for j in range(n):
        if j>0:
            print("*"*(j+1))
reverse_pyramid(10)
```
