## Sets
## Sets are built in data types in python,which are unordered and does not allow duplicates and are used to perform mathematical operations like unions,etc. 

```python
my_set = {1,2,3,4,5,6}
print(my_set)
print(type(my_set))
```

```python
lst_set = set([1,2,3,4,5,6,7,8])  ## converting list to set - can help in removing duplicates
lst_set.add(9)
lst_set.add(9) ## ignores duplicates..
print((lst_set))
lst_set.remove(4)  ##Remove element
print(lst_set)
lst_set.discard(11)  ## removes item if its present else doesnt throw error.
```

```python
## Pop Element

a = lst_set.pop()  ## Removes first element
print(a)
print(lst_set)
lst_set.clear() ## clears the set
print(lst_set)
```

```python
lst = [1,2,3,4,5,6,6,6,6,6]
set(lst)
## Set Membership test
print(6 in lst)
print(0 in lst)
```

```python
## Mathematical Sets
set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}

## Union Set
union_set = set1.union(set2)
print(union_set)

## Intersection Set
intersection_set = set1.intersection(set2)
print(intersection_set)  ##intersection elements

set1.intersection_update(set2)  ## intersection update basically updates the unique elements into one set.
print(set1) 
```

```python
set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}
## Difference Set
difference_set = set1.difference(set2)
print(difference_set)
print(set1)
set1.difference_update(set2)  ## difference update  = the not unique elements will be identified and updated to the set.
print(set1)
```

```python
## Symmetric Difference
set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}
sym_diff = set1.symmetric_difference(set2)
print(sym_diff)
## Symmetric Difference Update
set1.symmetric_difference_update(set2)
print(set1)
```

```python
seta = {1,2,3,4,5,6}
setb = {4,5,6}
## Subset -- contains all elements of setb in seta.
print(setb.issubset(seta))
## Superset   -- contains all elements from the other set
print(seta.issuperset(setb))
```

```python
##Counting Unique words
text = "counting Unique words are counting"
words = text.split()

ws = set(words)
print(ws)
print(len(ws))
```
