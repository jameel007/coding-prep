```python
## Right Angled Triangle II
## Output: ['  *', ' **', '***']
def right_angled_triangle(n):
    lst = []
    for i in range(n,0,-1):
        space = i-1 ## - 4
        star = n-i+1 ## - 
        row = " "*space + "*"* star
        lst.append(row)
    print(lst)
right_angled_triangle(4)
```

```python
##Inverted hollow right angle tri
def hollow_right_angled_triangel(n):
    lst = []
    for i in range(n,0,-1):
        if i == 1 or i==n:
            row="*"*i
        else:
            row = "*"+" "*(i-2)+"*"
        lst.append(row)
    print(lst)
hollow_right_angled_triangel(4)
```
