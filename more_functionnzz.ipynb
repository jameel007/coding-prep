## Unit Conversion

```python
## Unit Conversion
def unit_converter(temp,unit):
    if unit == "C": 
        return temp * 9/5 + 32  ## C to F
    elif unit =="F":
        return (temp-32) *5/9
    else:
        return None
print(unit_converter(50,"C"))
print(unit_converter(122,"F"))
print(unit_converter(75.2,""))
```

## Strong Password Checker

```python
def strong_password(password):
    if len(password)<8:
        return False
    elif not any(char.isdigit() for char in password):    ## Using keywords - any ,char,is , in 
        return False 
    elif not any(char.islower() for char in password):
        return False
    elif not any(char.isupper() for char in password):
        return False
    elif not any(char in '!@#$%^&*()' for char in password):
        return False
    else:
        return True
strong_password("ameeel!")
```

## Calculate Total Cost in Cart

```python
def shopping_cart(cart):
    total_cost = 0
    for item in cart:
        total_cost += item['Cost'] * item['Quantity']
    return total_cost


## Cart
cart = [ 
        {'Name':"Apple", 'Cost': 5 , 'Quantity': 10},
        {'Name':"Banana", 'Cost': 5 , 'Quantity': 10},
        {'Name':"Guava", 'Cost': 5 , 'Quantity': 10}
        ]    

print(shopping_cart(cart))
print (type(cart))
```

## Check for Palindrome


```python
def pal(s):
    s = s.lower().replace(" ","")
    return s==s[::-1]  ## here not using list just string slicing is being done to s with [].

print(pal("Racecar"))
```

## Factorial using Recursion

```python
def factorial(n):
    if n ==0:
        return 1
    else:
        print(n)
        return n * factorial(n-1)  
print(factorial(5))

""" The recursive function is a function call to itself 
    and whats happening her is even its calling till end the function 
    is waiting to each iteration until it resolves all of till end
    then it closes every paused function and resumes till end."""


"""Call 5 was waiting on:  return 1 * factorial(0)  →  1 * 1  =  1
Call 4 was waiting on:  return 2 * factorial(1)  →  2 * 1  =  2
Call 3 was waiting on:  return 3 * factorial(2)  →  3 * 2  =  6
Call 2 was waiting on:  return 4 * factorial(3)  →  4 * 6  =  24
Call 1 was waiting on:  return 5 * factorial(4)  →  5 * 24 = 120 """

"""GOING DOWN (calls pile up, nothing calculated yet):

factorial(5) → factorial(4) → factorial(3) → factorial(2) → factorial(1) 
→ factorial(0)       ↓
                    returns 1
                                                                         
COMING BACK UP (now the multiplication actually happens, in reverse order):
1 → 1*1=1 → 2*1=2 → 3*2=6 → 4*6=24 → 5*24=120"""
```

## A function to read a File and count the frequency of each word

```python
def count_words(file_path):
    word_count = {}
    with open(file_path,'r') as file:
        for line in file:
            words = line.split()
            for word in words:
                word = word.lower().strip(' ,-')
                word_count[word]=word_count.get(word,0)+1
    return word_count
file_path = "Learnt.txt"
print(count_words(file_path))
```

##Validate Email Address

```python
import re
def valid_email(email):
    pattern = r'^[a-zA-z0-9_.+-]+@[a-zA-z0-9]+\.[a-zA-z0-9.]'
    return re.match(pattern,email) is not None

print(valid_email('jameel@gmail'))
print(valid_email('jameel@gmail.co.in'))
```
