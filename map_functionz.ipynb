```python
## Map - The map() applies a given function to all items in a input list(or any other iterable) 
and return a map object (an iterator).This is particulaly useful for 
transforming data in a list comprehensively.  
```

```python
def sqr(x):
    return x**2
print(sqr(2))

a = [1,2,3,4,5,6,7,8,9]
b = list(map(sqr,a))
print(b)
```

```python
c = [1,2,3]
d = [4,5,6]
e = list(map(lambda x,y:x+y,c,d))  ## Headshot!
print(e)
```

```python
str_num = ['1','2','3']
num = list(map(int,str_num))  #3 string to numbers
print(num)
```

```python
f = ["Apple","Banana","Cat"]
g = list(map(str.upper,f))
print(type(f))
print(g)
```

```python
def get_name(x):
    return x['name']

x = [
    {'name':"Jameel",'Age':"30"},
    {'name': "Heena",'Age':40}
    ]

y = list(map(get_name,x))
print(y)
```
