## Filter functions

Filter returns the list of iterables only when the given condition is true rather than giving the entire list like map or lambda instead of tru and false it also gives the list!!

```python
def even(x):
    if x%2==0:
        return True
print(even(4))

lst = [1,2,3,4,5,6,7,8,9]

a = list(filter(even,lst)) ## goes through entire list and gives only the tru condition
b = list(map(even,lst))
c = list(map(lambda x:x%2==0,lst))
print(b)
print(c)
print(a)
```

```python
nums = [1,2,3,4,5,6,7,8,9]
gt_five = list(filter(lambda x:x>5,nums))
d = list(map(lambda x:x>5,nums))
print(d)
print(gt_five)

## Filter goes a step beyond to get the values 
# ...rather than just T or F of a condition.
```

```python
## Filter with lambda and multiple functions
nums = [1,2,3,4,5,6,7,8,9]
even_gt_five = list(filter(lambda x:x>5 and x%2==0,nums))
print(even_gt_five)
```

```python
## Check if age>25 in Dict

book =[
        {'name':"Jameel",'Age':30},
        {'name': "Heena",'Age':40}
    ]
def get_age(x):
    return x['Age']>30

list(filter(get_age,book))
```

The filter() is a powerful tool for creating iterators that filter items out of an iterable based on a condition. it is commonly used for data cleaning,filtering objects and removing unwanted elements from lists.
By mastering filter(), you can write more concise and efficient code for processing and manipulating collections in python.
