```python
## Odd or Even?
num = int(input("Enter a number:"))
if num == 0:
    print("The num you entered is neither even or odd")
elif num % 2 == 0:
    print("The num you have entered is Even")
else:
    print("The num you have entered is Odd")
```

```python
## Break , Continue and Pass
for i in range(10):
    
    if i == 4:
        pass
        print("This is 4") ## Even with Pass below line executes
    elif i == 5:
        continue ## this skips the iteration but attempts for another iteration
        print("This is 5")
    elif i == 8:
        break
        print("Try to print this: this shouldnt print") ##Use of break - it exits the loop instantly not to return again breaking the loop
    print(i)
```

```python
## Factorial - n = n * n-1 * n-2
n = int(input("Enter a number to calculate factorial:"))
f = 1
for i in range(n):
        x = n -i
        f = f*x
print(f)
```
