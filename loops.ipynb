## Control flow 
## Break
## Break exxits the loop permanently

```python
for i in range(10):
    if i==10:
        break    ##stops the entire loop once and for all no more iteration
    print("Im still inside the loop",i)  
print("The loop broke here at the value",i)          ##print("Here i am", i)
```

## Continue Statement
## The continue skips the current iteration and continues with the next.

```python
for i in range(10):
    if i%2==0:
        continue  ## skips on this particular iteration matching the condition and continues the loop
    print(i)
```

## Pass Statement
## The pass statement is a null operation; it does nothing.

```python
for i in range(5):
    if i == 3:
        print("I am going to pass this number",i)
        pass ## Does nothing at all.
    print(i)
```

## Pass - does nothing - more like placeholder and used it class by not breaking functionality; class needs identation so we use pass for empty class; Pass is useful in exception like a resume from BP.

## Nested Loops

```python
for i in range(3):
    for j in range(2):
        print(f"j is {j} and i is {i}")  
```

```python
## Sum of first 10 natural numbers 
## while loop
n = 10
sum = 0
count = 1 
while count <= n:
    sum = sum + count
    count = count + 1
print("Sum of first 10 natural numbers is",sum)
```

```python
## Sum of first 10 natural numbers 
##For loop 
## Executed at first go - Alhamdulillah!!

sum = 0
for i in range(11):
    sum = sum + i
print("Sum of first 10 natural numbers is",sum)
```

```python
## Print Prime numbers between 1 and 100
count = 0
for i in range(1,101):
    if i > 1:
        for n in range(2, i):
            if i % n == 0:
                break
        else:
            count = count +1
            print(i)
print("Total number of prime numbers found between 1 to 100 is",count)
```
