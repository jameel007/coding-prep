## Lambda

A lambda function is a single expression function using the keyword lambda.It can have any number of argument but only one expression.They are used for short operations or as arguments to higher-order functions.

```python
def add(a,b):
    return a+b
print(add(2,3))

## Syntax = lambda arguments: expression
c = lambda a,b:a+b
d = lambda e,f,g,h,i,j:e*f*g*h*i*j
print(type(c))
print(c(1,2))
print(d(1,2,3,4,5,6))
```

```python
## Map  - applies a function to all items in the list/..
nums = [1,2,3,4,5,6,7,8,9]
def square(nums):
    return nums**2
print(square(2))

## to loop every number instead we can use lambda
list(map(lambda x:x**2,nums))    ## - dammmnnn
```
